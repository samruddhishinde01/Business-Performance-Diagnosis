```python
df.dropna()
```

```python
#Total Revenue, Cost and Profit
total_revenue=df["Billable_Revenue"].sum()
total_cost=df["Total_Transaction_Cost"].sum()
total_profit=df["Transaction_Profit"].sum()

print("Total Revenue:", round(total_revenue, 2))
print("Total Cost:", round(total_cost, 2))
print("Total Profit:", round(total_profit, 2))
```

```python
#Overall Profit Margin
profit_margin=(total_profit/total_revenue)*100
print("Profit_Margin :",round(profit_margin,2),"%")
```

```python
#Monthly Revenue and Profit
monthly_analysis=df.groupby("Month")[["Billable_Revenue","Transaction_Profit"]].sum()
monthly_analysis = monthly_analysis / 10000000

print(monthly_analysis)
```

```python
#Monthly Profit Margin
monthly_analysis = df.groupby("Month").agg(
    Revenue=("Billable_Revenue", "sum"),
    Profit=("Transaction_Profit", "sum")
)

monthly_analysis["Profit_Margin"] = (
    monthly_analysis["Profit"] / monthly_analysis["Revenue"] * 100
)

print(monthly_analysis.round(2))
```

```python
#Service Line Performance
service_analysis = df.groupby("Service_Line").agg(
    Revenue=("Billable_Revenue", "sum"),
    Profit=("Transaction_Profit", "sum")
)

service_analysis["Profit_Margin"] = (
    service_analysis["Profit"] / service_analysis["Revenue"] * 100
)

print(service_analysis.sort_values("Revenue", ascending=False).round(2))
```

```python
#Client Industry Analysis
industry_analysis = df.groupby("Client_Industry").agg(
    Revenue=("Billable_Revenue", "sum"),
    Profit=("Transaction_Profit", "sum")
)

industry_analysis["Profit_Margin"] = (
    industry_analysis["Profit"] / industry_analysis["Revenue"] * 100
)

print(industry_analysis.round(2))
```

```python
#Cost Component Analysis
cost_analysis = df[[
    "Labor_Cost",
    "Technology_Cost",
    "Operational_Cost",
    "Other_Direct_Cost",
    "Allocated_Overhead"
]].sum()

print(cost_analysis)
```

```python
#Billable vs Non-Billable Hours
print("Billable Hours:", df["Billable_Hours"].sum())

print("Non-Billable Hours:", df["Non_Billable_Hours"].sum())

print("Total Hours:", df["Hours_Worked"].sum())
```

```python
#Department Utilization
department_analysis = df.groupby("Employee_Department")[
    "Resource_Utilization_Percentage"
].mean()

print(department_analysis.round(2))
```

```python
#Project Delivery Analysis
delivery_analysis = p_df.groupby("Delivery_Status").size()

print(delivery_analysis)
```

```python
#Project Type Analysis
project_type_analysis = df.groupby("Project_Type").agg(
    Revenue=("Billable_Revenue", "sum"),
    Profit=("Transaction_Profit", "sum")
)

print(project_type_analysis.round(2))
```

```python
#Final Business Insights
print("Highest Revenue Service Line:")
print(service_analysis["Revenue"].idxmax())

print("Lowest Profit Margin Service Line:")
print(service_analysis["Profit_Margin"].idxmin())

print("Highest Cost Component:")
print(cost_analysis.idxmax())

print("Lowest Utilization Department:")
print(department_analysis.idxmin())

print("Highest Revenue Project Type:")
print(project_type_analysis["Revenue"].idxmax())
```
